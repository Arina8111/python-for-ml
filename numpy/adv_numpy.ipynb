NUMPY ARRAY VS PYTHON LISTS

```python
import time
a=[i for i in range(10000000)]
b=[i for i in range(10000000,20000000)]

c=[]
start=time.time()
for i in range(len(a)):
    c.append(a[i]+b[i])
print(time.time()-start)
```

```python
a=np.arange(1000000)
b=np.arange(1000000,2000000)
start=time.time()
c=a+b
print(time.time()-start)
```

```python
# memory
a=[i for i in range(10000000)]
import sys
sys.getsizeof(a)
```

```python
a=np.arange(1000000,dtype=np.int32)
sys.getsizeof(a)
```
